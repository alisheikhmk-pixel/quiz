```python

import numpy as np

def softmax(x):
    x = x - np.max(x)

    exp_values = np.exp(x)

    probabilities = exp_values / np.sum(exp_values)

    return probabilities


x = np.array([2.0, 1.0, 0.1])

output = softmax(x)

print("Softmax Output:", output)
print("Sum of Probabilities:", np.sum(output))
```

### Self-Attention Forward Pass

Self-attention allows the model to focus on different parts of the input sequence. The steps for the scaled dot-product attention forward pass are:
1. Project the input sequence $X$ into Queries ($Q$), Keys ($K$), and Values ($V$) using weight matrices $W_Q$, $W_K$, and $W_V$.
2. Compute the attention scores: $\text{Scores} = QK^T$.
3. Scale the scores by the square root of the key dimension ($\sqrt{d_k}$).
4. Apply Softmax to get the attention weights.
5. Multiply the attention weights by the Values ($V$) to get the final output.

```python
def self_attention(X, W_Q, W_K, W_V):
    # Get Q, K, V matrices
    Q = np.dot(X, W_Q)
    K = np.dot(X, W_K)
    V = np.dot(X, W_V)

    # Size of key vectors
    d_k = K.shape[-1]

    # Calculate raw match scores
    scores = np.dot(Q, K.T) / np.sqrt(d_k)

    # Convert scores to percentages (softmax)
    scores_stabilized = scores - np.max(scores, axis=-1, keepdims=True)
    exp_scores = np.exp(scores_stabilized)
    attention_weights = exp_scores / np.sum(exp_scores, axis=-1, keepdims=True)

    # Combine weights with V
    output = np.dot(attention_weights, V)

    return output, attention_weights

# Random seed for same results
np.random.seed(42)

# Input data (3 words, 4 features each)
X = np.random.randn(3, 4)

#Dimension of key
d_k = 2

# Projection weights
W_Q = np.random.randn(4, d_k)
W_K = np.random.randn(4, d_k)
W_V = np.random.randn(4, d_k)

# Run self-attention
output, weights = self_attention(X, W_Q, W_K, W_V)
```
