```python
def self_attention(X, W_Q, W_K, W_V):
    # Get Q, K, V matrices
    Q = np.dot(X, W_Q)
    K = np.dot(X, W_K)
    V = np.dot(X, W_V)

    # Size of key vectors
    d_k = K.shape[-1]

    # Calculate raw match scores
    scores = np.dot(Q, K.T) / np.sqrt(d_k)

    # Convert scores to percentages (softmax)
    scores_stabilized = scores - np.max(scores, axis=-1, keepdims=True)
    exp_scores = np.exp(scores_stabilized)
    attention_weights = exp_scores / np.sum(exp_scores, axis=-1, keepdims=True)

   # Combine weights with V
    output = np.dot(attention_weights, V)

    return output, attention_weights

# Random seed for same results
np.random.seed(42)

# Input data (3 words, 4 features each)
X = np.random.randn(3, 4)

#Dimension of key
d_k = 2

# Projection weights
W_Q = np.random.randn(4, d_k)
W_K = np.random.randn(4, d_k)
W_V = np.random.randn(4, d_k)

# Run self-attention
output, weights = self_attention(X, W_Q, W_K, W_V)
```
